---

## 2. Basic Concept of FFT

FFT stands for **Fast Fourier Transform**.

FFT is used to analyze the frequency components present in a sampled signal.

In simple words:

> FFT helps us find which frequencies are present in a signal.

In Day 3, we mainly observed the signal in the **time domain**.

In this task, we analyze the signal in the **frequency domain**.

### Time Domain

The time-domain representation tells us:

> How does the signal change with time?

- X-axis → Time
- Y-axis → Signal amplitude

### Frequency Domain

The frequency-domain representation tells us:

> Which frequency components are present in the signal?

- X-axis → Frequency
- Y-axis → Magnitude

The basic process is:

```text
Time-Domain Signal
        ↓
       FFT
        ↓
Frequency-Domain Information
        ↓
Frequency Spectrum
        ↓
Frequency Peaks

````python
import numpy as np
import matplotlib.pyplot as plt
````

````python
f=10
f2=20
fs=100
t = np.arange(0, 1, 1/fs)
x = np.sin(2 * np.pi * f * t)+np.sin(2 * np.pi * f2* t)
y=np.fft.fft(x)
magnitude=np.abs(y)
frequency = np.fft.fftfreq(len(x),1/fs)
plt.plot(frequency,magnitude)
plt.xlabel("Frequency (Hz)")
plt.ylabel("Magnitude")
plt.title("FFT Spectrum - Known Synthetic Signal")
plt.show()
````

# Test 2 — Unknown Synthetic Signal Analysis Using FFT

````python
f=np.random.randint(5,25)
fs=100
t=np.arange(0,1,1/fs)
x=np.sin(2*np.pi*f*t)
y=np.fft.fft(x)
magnitude=np.abs(y)
frequency = np.fft.fftfreq(len(x),1/fs)
plt.plot(frequency,magnitude)
plt.xlabel("Frequency (Hz)")
plt.ylabel("Magnitude")
plt.title("FFT Spectrum - Unknown Synthetic Signal")
plt.show()
print("Random Frequency:",f)
````

## 5. Synthetic Signal Generation

A sine-wave signal was generated using the selected frequency and the time array.

The basic sine-wave equation used was:

**x(t) = sin(2πft)**

where:

- `x(t)` = signal value
- `f` = frequency
- `t` = time
- `π` = mathematical constant

The generated sine wave was used as the synthetic test signal.

## 7. Applying FFT

The FFT was applied to the synthetic signal using NumPy.

The FFT converts the signal from the time domain into the frequency domain.

The FFT output contains complex values.

Therefore, the absolute value of the FFT output was calculated to obtain the magnitude.

---

## 8. FFT Magnitude

The magnitude of the FFT was calculated using:

`np.abs()`

The magnitude represents the strength of the frequency components.

A strong frequency component appears as a peak in the FFT spectrum.

Therefore:

**Higher peak → Stronger frequency component**

---

## 9. Frequency Axis

The frequency axis was created using:

`np.fft.fftfreq()`

The number of samples and sampling frequency were used to calculate the corresponding frequency values.

This allows the FFT result to be represented in Hertz.

---

## 10. Frequency Spectrum

The FFT spectrum was plotted using Matplotlib.

### X-axis

**Frequency (Hz)**

### Y-axis

**Magnitude**

The spectrum shows the frequency content of the synthetic signal.

---

## 11. Observation

After running the FFT, a clear peak was observed in the frequency spectrum.

The generated random frequency was verified as:

**11 Hz**

The corresponding positive-frequency peak represents the 11 Hz component.

A corresponding negative-frequency component can also appear because the FFT of a real-valued signal contains positive and negative frequency components.

## 12. Understanding the Graph

The graph can be interpreted as follows:

- X-axis tells us the frequency.
- Y-axis tells us the magnitude.
- A peak indicates the presence of a frequency component.
- The position of the peak gives the approximate frequency.

Therefore:

**Peak position → Frequency component**
