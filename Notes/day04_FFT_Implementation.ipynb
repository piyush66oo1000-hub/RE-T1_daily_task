# Test 1 — Known Synthetic Signal FFT Validation

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
f=10
f2=20
fs=100
t = np.arange(0, 1, 1/fs)
x = np.sin(2 * np.pi * f * t)+np.sin(2 * np.pi * f2* t)
y=np.fft.fft(x)
magnitude=np.abs(y)
frequency = np.fft.fftfreq(len(x),1/fs)
plt.plot(frequency,magnitude)
plt.xlabel("Frequency (Hz)")
plt.ylabel("Magnitude")
plt.title("FFT Spectrum - Known Synthetic Signal")
plt.show()
```

## Test 1 — Explanation and Observation

### 1. Objective

The purpose of this test is to verify the FFT implementation using a synthetic signal whose frequency components are already known.

The signal contains two known frequency components:

- **10 Hz**
- **20 Hz**

The FFT should therefore show peaks around these frequencies.

---

### 2. Sampling Frequency

The sampling frequency is:

**fs = 100 Hz**

This means the signal is sampled 100 times per second.

The sampling interval is:

**1 / fs = 0.01 seconds**

---

### 3. Time Array

The time array represents the points at which the signal is sampled.

The signal duration is 1 second, with a sampling interval of 0.01 seconds.

---

### 4. Synthetic Signal

Two sine waves are generated:

- First component = 10 Hz
- Second component = 20 Hz

The two components are added together to form one synthetic signal.

The signal can be represented as:

**x(t) = sin(2πft) + sin(2πf₂t)**

---

### 5. Applying FFT

FFT converts the signal from the time domain into the frequency domain.

The FFT output contains complex values.

The magnitude is calculated using:

`np.abs()`

This makes it easier to observe the strength of the frequency components.

---

### 6. Frequency Axis

`np.fft.fftfreq()` is used to create the frequency axis.

The sampling frequency and number of samples are used to convert FFT bins into frequency values in Hertz.

---

### 7. Observation

The FFT spectrum shows peaks around:

- **+10 Hz**
- **+20 Hz**

Corresponding negative-frequency components also appear around:

- **-10 Hz**
- **-20 Hz**

The negative-frequency components are a normal property of the FFT of a real-valued signal.

---

### 8. Test Result

The expected frequency components were correctly visible in the FFT spectrum.

Therefore, the FFT implementation was successfully validated.

---

### 9. Key Learning

**Time domain → FFT → Frequency domain**

FFT helps identify the frequency components present in a digital signal.

# Test 2 — Random Synthetic Signal Analysis Using FFT

```python
f=np.random.randint(5,25)
fs=100
t=np.arange(0,1,1/fs)
x=np.sin(2*np.pi*f*t)
y=np.fft.fft(x)
magnitude=np.abs(y)
frequency = np.fft.fftfreq(len(x),1/fs)
plt.plot(frequency,magnitude)
plt.xlabel("Frequency (Hz)")
plt.ylabel("Magnitude")
plt.title("FFT Spectrum - Unknown Synthetic Signal")
plt.show()
print("Random Frequency:",f)
```

## Test 2 — Explanation and Observation

### 1. Objective

The objective of this test is to analyze a synthetic signal using FFT and identify its frequency component from the frequency spectrum.

Unlike Test 1, the frequency is automatically selected using NumPy rather than being manually assigned.

---

### 2. Sampling Setup

The sampling frequency is:

**fs = 100 Hz**

Therefore:

- Samples per second = 100
- Sampling interval = 1 / 100 = **0.01 seconds**
- Signal duration = **1 second**

---

### 3. Random Frequency Generation

NumPy random number generation is used to select the frequency.

The function:

`np.random.randint(5, 25)`

selects an integer frequency from the selected range.

A fixed random seed is used so that the experiment is reproducible.

For this run, the generated frequency is:

**11 Hz**

---

### 4. Synthetic Signal Generation

A sine-wave signal is generated using the selected frequency.

The equation is:

**x(t) = sin(2πft)**

where:

- `x(t)` = signal value
- `f` = generated frequency
- `t` = time

The resulting signal is used as the synthetic test signal.

---

### 5. Applying FFT

FFT converts the time-domain signal into the frequency domain.

The FFT output contains complex values.

The magnitude is calculated using:

`np.abs()`

The magnitude represents the strength of the frequency component.

---

### 6. Frequency Axis

The frequency axis is generated using:

`np.fft.fftfreq()`

The sampling frequency and number of samples are used to calculate the frequency values in Hertz.

---

### 7. Frequency Spectrum

The spectrum is plotted with:

- **X-axis → Frequency (Hz)**
- **Y-axis → Magnitude**

The important information is obtained from the location of the major peak.

---

### 8. Observation

The FFT spectrum shows a strong positive-frequency peak around:

**11 Hz**

A corresponding negative-frequency peak also appears around:

**-11 Hz**

This is expected for the FFT of a real-valued sine signal.

---

### 9. Verification

The generated frequency was temporarily printed for verification.

The output was:

**Generated Frequency: 11 Hz**

The FFT spectrum also shows the corresponding peak around 11 Hz.

This confirms that the FFT correctly represents the frequency content of the generated synthetic signal.
