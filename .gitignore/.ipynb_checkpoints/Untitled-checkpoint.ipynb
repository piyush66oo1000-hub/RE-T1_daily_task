# DSP Fundamentals — Day 3

## Objective

To understand basic Digital Signal Processing (DSP)
concepts and discrete signal models using Python and NumPy.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.array([7, 14, 21, 28])
n = np.arange(4)

print("Sample numbers:", n)
print("Signal values:", x)
```

## 1. Discrete Signal Representation

A discrete signal is represented using a sequence of samples.

Here, `n` represents the sample number (index), while `x[n]`
represents the signal value at that sample.

For example:

- n = [0, 1, 2, 3]
- x[n] = [7, 14, 21, 28]

Therefore:

- x[0] = 7
- x[1] = 14
- x[2] = 21
- x[3] = 28

```python
plt.stem(n, x)
plt.xlabel("Sample Number (n)")
plt.ylabel("Signal Value x[n]")
plt.title("Discrete Signal")
plt.grid()
plt.show()
```

## 2. Sampling a Sine Signal

A sine signal can be represented digitally by taking samples at
regular time intervals.

The signal frequency is represented by `f`, while `fs` represents
the sampling frequency.

In this example:

- f = 10 Hz
- fs = 100 Hz
- Sampling interval = 1/fs = 0.01 seconds

```python
f = 10
fs = 100

t = np.arange(0, 1, 1/fs)

x = np.sin(2 * np.pi * f * t)

print("Number of samples:", len(t))
print("First 10 time samples:", t[:10])
```

```python
plt.plot(t, x)
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.title("Sampled Sine Signal")
plt.grid()
plt.show()
```

## 3. Observation

The generated signal has a frequency of 10 Hz and is sampled at
100 samples per second.

The time interval between two consecutive samples is:

Sampling interval = 1/fs = 1/100 = 0.01 seconds

Since the signal frequency is 10 Hz, approximately 10 cycles
are present in one second.

The graph shows the sampled sine signal over the selected
one-second time interval.

```python
fs2 = 20

t2 = np.arange(0, 1, 1/fs2)

x2 = np.sin(2 * np.pi * f * t2)

plt.stem(t2, x2)
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.title("Sine Signal with Lower Sampling Rate")
plt.grid()
plt.show()
```

## 4. Effect of Sampling Rate

The signal frequency is kept constant at 10 Hz.

Two different sampling rates are compared:

- First case: fs = 100 Hz
- Second case: fs = 20 Hz

When the sampling rate is reduced, fewer samples are taken
during the same time interval.

Therefore, the discrete representation contains fewer samples.

## 5. Discrete Signal and Sampled Sine Signal

A discrete signal is represented as a sequence of individual
sample values indexed by `n`.

For the manually defined signal:

- `n` represents the sample index.
- `x[n]` represents the signal value at that index.

For the sampled sine signal:

- `t` represents the sampling time points.
- `x` contains the calculated signal values at those time points.

Therefore, sampling converts a signal into a sequence of
numerical values that can be processed digitally.

```python
print("Signal frequency:", f, "Hz")
print("Sampling frequency:", fs, "Hz")
print("Sampling interval:", 1/fs, "seconds")
print("Number of samples:", len(x))
```

## 6. Final Observations

1. A discrete signal consists of individual samples.
2. `n` represents the sample index.
3. `x[n]` represents the signal value at a sample.
4. `f` represents the signal frequency.
5. `fs` represents the sampling frequency.
6. The sampling interval is calculated as `1/fs`.
7. Increasing the sampling rate provides more samples over the same time interval.
8. NumPy was used to create and process signal samples.
9. Matplotlib was used to visualize the signals.
