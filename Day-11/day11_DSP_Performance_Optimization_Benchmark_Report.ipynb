## DSP Performance Optimization: Benchmark Report
### Objective
Compare a Python loop with NumPy vectorization using the same synthetic input. Measure execution times, verify output correctness, and calculate speedup.

**Scope:** Local, software-only benchmark using synthetic data. Timings depend on your computer and system load. Run the notebook to get actual measurements; no timing results are invented. This does not measure RF hardware or real RF captures.

## 1. Imports and synthetic input
Run this cell first.

```python
import time
import statistics
import numpy as np
import pandas as pd

N = 100_000
REPEATS = 5
signal = np.arange(N, dtype=np.float64)

print("Samples:", len(signal))
print("Repeated measurements per method:", REPEATS)
```

## 2. Define methods and check correctness
Both methods multiply every sample by 2.

```python
def loop_method(data):
    result = []
    for x in data:
        result.append(2 * x)
    return result

def numpy_method(data):
    return 2 * data

loop_result = loop_method(signal)
numpy_result = numpy_method(signal)
outputs_match = np.allclose(loop_result, numpy_result)

print("Outputs match:", outputs_match)
assert outputs_match, "The methods produced different results." 
```

## 3. Repeated benchmark
A warm-up run is excluded from the recorded timings.

```python
def benchmark(function, data, repeats=REPEATS):
    function(data)  # warm-up
    run_times = []
    for _ in range(repeats):
        start = time.perf_counter()
        function(data)
        end = time.perf_counter()
        run_times.append(end - start)
    return run_times

loop_times = benchmark(loop_method, signal)
numpy_times = benchmark(numpy_method, signal)

print("Loop run times (seconds):", [round(x, 6) for x in loop_times])
print("NumPy run times (seconds):", [round(x, 6) for x in numpy_times])
loop_mean = statistics.mean(loop_times)
numpy_mean = statistics.mean(numpy_times)
print("Loop mean (seconds):", round(loop_mean, 6))
print("NumPy mean (seconds):", round(numpy_mean, 6))
```

## 4. Speedup and time reduction
Speedup above 1 means NumPy was faster in this measured run.

```python
speedup = loop_mean / numpy_mean if numpy_mean > 0 else float("nan")
reduction = ((loop_mean - numpy_mean) / loop_mean * 100) if loop_mean > 0 else float("nan")

print(f"Speedup (loop mean / NumPy mean): {speedup:.3f}x")
print(f"Execution-time reduction: {reduction:.2f}%")
print("Output correctness:", outputs_match)
```

## 5. Summary table
This table displays measurements from the cells above.

```python
summary = pd.DataFrame({
    "Method": ["Python loop", "NumPy vectorization"],
    "Mean time (seconds)": [loop_mean, numpy_mean],
    "Fastest run (seconds)": [min(loop_times), min(numpy_times)],
    "Slowest run (seconds)": [max(loop_times), max(numpy_times)],
    "Output verified": [outputs_match, outputs_match],
})
summary
```
