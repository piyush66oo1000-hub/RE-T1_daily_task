# Day 5 — Spectrogram Generation

**Objective:** Generate a synthetic signal and visualize its frequency content over time using a spectrogram.



```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
f1=50
f2=150
fs=1000
t = np.arange(0, 2, 1/fs)

signal = np.concatenate([
    np.sin(2 * np.pi * f1 * t[:1000]),
    np.sin(2 * np.pi * f2 * t[1000:])
])
```

```python

plt.figure(figsize=(12, 6))

plt.specgram(
    signal,
    NFFT=256,
    Fs=fs,
    noverlap=192,
    cmap="viridis",
    scale="dB"
)

plt.ylim(0, 250)
plt.xlim(0, 2)

plt.title("Spectrogram Analysis", fontsize=15)
plt.xlabel("Time (seconds)", fontsize=12)
plt.ylabel("Frequency (Hz)", fontsize=12)

plt.colorbar(label="Power (dB)")
plt.grid(alpha=0.2)
plt.tight_layout()

plt.show()
```
